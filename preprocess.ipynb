```python
import csv

with open("data/regions.csv") as csvf:  
    regions = list(csv.DictReader(csvf))
    company_to_region_id = {}
    address_to_region_id = []
    
    for row in regions:
        if row["address"] not in address_to_region_id:
            address_to_region_id.append(row["address"])
    
    print(address_to_region_id)
    
    for row in regions:
        company_to_region_id[row["id"]] = str(address_to_region_id.index(row["address"]) + 1)
    
    print(company_to_region_id)
            
```

```python
schools_df = pd.read_csv('data/schools.csv')
founders_df = pd.read_csv('data/founders.csv')

founder_hnids = founders_df['hnid'].unique()
schools_df = schools_df[schools_df['hnid'].isin(founder_hnids)]

schools_df.to_csv('data/schools.csv', index=False)
```

```python
df = pd.read_csv('data/schools.csv')
print(df.shape)
df.head()
```

```python
unique_counts = df.nunique()
print(unique_counts)
```

```python
nan_count_per_column = df.isna().sum()
print(nan_count_per_column)
```

```python
counts_of_unique_values = df['hnid'].value_counts()
print(counts_of_unique_values)
```

```python
null_field_mask = df['field_of_study'].isnull()

non_null_field = df[~null_field_mask].groupby(['hnid', 'school', 'field_of_study']).agg({'year': 'max'}).reset_index()

null_field = df[null_field_mask].groupby(['hnid', 'school']).agg({'year': 'max', 'field_of_study': 'first'}).reset_index()

preprocessed_df = pd.concat([non_null_field, null_field], ignore_index=True)
```

```python
preprocessed_df.drop_duplicates(subset=['hnid', 'school', 'year'], inplace=True)
```

```python
counts_of_unique_values = preprocessed_df['hnid'].value_counts()
print(counts_of_unique_values)
```

```python
preprocessed_df.to_csv('data/schools.csv', index=False)
```
